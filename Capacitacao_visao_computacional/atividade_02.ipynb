```python
import os
from pathlib import Path

import cv2
import matplotlib.pyplot as plt
import numpy as np
```

```python
BASE_DIR = Path.cwd()
imagens_o = BASE_DIR / "data" / "letras_O&U" / "letra_o"
imagens_u = BASE_DIR / "data" / "letras_O&U" / "letra_u"
```

```python
img_o1 = cv2.imread(str(f"{imagens_o}/o2.jpg"))
img_o2 = cv2.imread(str(f"{imagens_o}/o3.jpg"))
img_u1 = cv2.imread(str(f"{imagens_u}/u2.jpg"))
img_u2 = cv2.imread(str(f"{imagens_u}/u5.jpg"))
```

```python
def filtro_media(img, kernels):
    imagem_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
    blur_media = cv2.blur(imagem_rgb, kernels)
    plt.figure(figsize=(10,10))
    plt.title(f'Filtro da Média com kernels {kernels[0]} x {kernels[1]}')
    plt.imshow(blur_media)
    plt.axis('off')
    plt.show()
```

```python
filtro_media(img_o1, (3, 3))
```

```python
filtro_media(img_o1, (9, 9))
```

```python
def filtro_gaussiano(img, kernels, sigma):
    imagem_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
    gaussiano = cv2.GaussianBlur(imagem_rgb, kernels, sigma)
    plt.figure(figsize=(10,10))
    plt.title(f'Filtro Gaussiano com kernels {kernels[0]} x {kernels[1]} sigma = {sigma}')
    plt.imshow(gaussiano)
    plt.axis('off')
    plt.show()
```

```python
filtro_gaussiano(img_u1, (9, 9), 1)
```

```python
filtro_gaussiano(img_u2, (9, 9), 3)
```

```python
def filtro_mediana(img, kernels):
    imagem_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
    mediana = cv2.medianBlur(imagem_rgb, kernels)
    plt.figure(figsize=(10,10))
    plt.title(f'Filtro  da Mediana com kernels {kernels} x {kernels}')
    plt.imshow(mediana)
    plt.axis('off')
    plt.show()
```

```python
filtro_mediana(img_o2, (3))
```

```python
filtro_mediana(img_o2, (5))
```
